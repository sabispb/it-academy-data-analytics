### 1. Consulta l’API pública [JSONPlaceholder](https://jsonplaceholder.typicode.com/) utilitzant el mètode GET per obtenir:

- Llista de publicacions (/posts)
- Llista d’usuaris (/users)
- Llista de tasques (/todos)

```python
BASE_URL = "https://jsonplaceholder.typicode.com"

def get_results(endpoint: str):
    try:
        response = requests.get(f"{BASE_URL}/{endpoint}")
        response.raise_for_status()
        return response.json(), response.status_code
    except HTTPError as http_err:
        print(f"HTTP error for {endpoint}: {http_err}")
    except Exception as err:
        print(f"Other error for {endpoint}: {err}")
    return None
```

```python
posts, posts_status = get_results("posts")
users, users_status = get_results("users")
todos, todos_status = get_results("todos")
```

```python
pp.pprint(posts)
```

```python
pp.pprint(users)
```

```python
pp.pprint(todos)
```

### 4. Fes una petició POST per crear una nova publicació fictícia. Inclou el títol, el cos del missatge i un userId.

- Mostra la resposta JSON.
- Mostra el codi d'estat.

```python
url = "https://jsonplaceholder.typicode.com/posts"
              
data={
    'body': 'hola, em dic Sabina',
    'title': 'hola',
    'userId': 56
}

response = requests.post(url, json=data)
print(f"Codi d'estat: {response.status_code}\n")
print(f"Resposta JSON:\n{response.json()}")
```

### 5. Fes una petició PATCH per modificar parcialment una publicació existent.

- Mostra la resposta JSON.
- Mostra el codi d'estat.

```python
url = "https://jsonplaceholder.typicode.com/posts/101"
              
data={
    'body': 'ADEU',
}

response = requests.patch(url, json=data)
print(f"Codi d'estat: {response.status_code}\n")
print(f"Resposta JSON:\n{response.json()}")
```

```python
# Consultar posts del dataset original
url = "https://jsonplaceholder.typicode.com/posts/101"

response = requests.get(url)
print(f"\nGET status: {response.status_code}")
print(response.json())
```
