### 1. Consulta l’API pública [JSONPlaceholder](https://jsonplaceholder.typicode.com/) utilitzant el mètode GET per obtenir:

- Llista de publicacions (/posts)
- Llista d’usuaris (/users)
- Llista de tasques (/todos)

```python
BASE_URL = "https://jsonplaceholder.typicode.com"

def get_results(endpoint: str):
    try:
        response = requests.get(f"{BASE_URL}/{endpoint}")
        response.raise_for_status()
        return response.json(), response.status_code
    except HTTPError as http_err:
        print(f"HTTP error for {endpoint}: {http_err}")
    except Exception as err:
        print(f"Other error for {endpoint}: {err}")
    return None
```

```python
posts, posts_status = get_results("posts")
users, users_status = get_results("users")
todos, todos_status = get_results("todos")
```

```python
pp.pprint(posts)
```

```python
pp.pprint(users)
```

```python
pp.pprint(todos)
```

From construct the query [documentation](https://open.fda.gov/apis/query-parameters/):

**Query parameters**
The API supports five query parameters. The basic building block of queries is the search parameter. Use it to “filter” requests to the API by looking in specific fields for matches. Each endpoint has its own unique fields that can be searched.

- search: What to search for, in which fields. If you don’t specify a field to search, the API will search in every field.
- sort: Sort the results of the search by the specified field in ascending or descending order by using the :asc or :desc modifier.
- count: Count the number of unique values of a certain field, for all the records that matched the search parameter. By default, the API returns the 1000 most frequent values.
- limit: Return up to this number of records that match the search parameter. Currently, the largest allowed value for the limit parameter is 1000.
- skip: Skip this number of records that match the search parameter, then return the matching records that follow. Use in combination with limit to paginate results.
Currently, the largest allowed value for the skip parameter is 25000. See Paging if you require paging through larger result sets.

### 3. Fes una petició GET senzilla:

Mostra el codi d'estat de la resposta.
Imprimeix de forma clara alguns dels camps de la resposta JSON.

```python
response = requests.get("https://api.fda.gov/drug/event.json")
print(f"\nGET status: {response.status_code}")
pp.pprint(response.json())
```

```python
response = requests.get('https://api.fda.gov/drug/event.json?search=occurcountry:"US"&sort=transmissiondate:desc&limit=3')
print(f"\nGET status: {response.status_code}")
pp.pprint(response.json())

# search=occurcountry:"Spain"&sort=transmissiondate:desc&limit=10
```

From [documentation](https://open.fda.gov/apis/drug/event/how-to-use-the-endpoint/) on how to use an endpoint:
Some key pointers
- An openFDA API query always begins with the base endpoint, which in this case is: https://api.fda.gov/drug/event.json
- Searches have a special syntax: search=field:term
- Unless otherwise specified, the API will return only one matching record for a search. You can specify the number of records to be returned by using the limit parameter. The maximum limit allowed is 1000 for any single API call. If no limit is set, the API will return one matching record.

```python
url = 'https://api.fda.gov/drug/event.json?search=patient.drug.openfda.pharm_class_epc:"nonsteroidal+anti-inflammatory+drug"&count=patient.reaction.reactionmeddrapt.exact'

response = requests.get(url)
print(f"\nGET status: {response.status_code}")
pp.pprint(response.json())
```

## Nivell 3

[API d'Open Data Barcelona](https://opendata-ajuntament.barcelona.cat/ca/desenvolupadors#APIS)

### 1. Utilitza l’API d’Open Data BCN per:

Cercar un dataset del teu interès mitjançant package_search i package_show.

```python
url = 'https://opendata-ajuntament.barcelona.cat/data/api/action/'
method = 'package_search'

response = requests.get(url+method)
print(f"\nGET status: {response.status_code}")

data = response.json()

first_dataset = data["result"]["results"][0]

pp.pprint(first_dataset)
```

```python
print(first_dataset.keys())
```

```python
print(first_dataset["title"])
```

```python
datasets = data["result"]["results"]

for d in datasets:
    print(d["title"])

print('----')
dataset_id = datasets[6]["id"]
print(datasets[6]["title"])
```

```python
url = 'https://opendata-ajuntament.barcelona.cat/data/api/action/'
method = 'package_show'

params = {"id": dataset_id}

response = requests.get(url+method, params=params)
data = response.json()
print(f"\nGET status: {response.status_code}")
pp.pprint(data["result"])
```

### 2. Dels resultats obtinguts, selecciona’n un que tingui recursos disponibles en CSV o JSON.

```python
for d in datasets:
    formats = set()  # per evitar duplicats
    
    for r in d["resources"]:
        fmt = r["format"]
        if fmt:  # evitar buits
            formats.add(fmt.upper())
    
    print(f"{d['title']} → {', '.join(formats)}")
```

### 3. Amb el resource_id del recurs seleccionat, realitza una consulta mitjançant datastore_search per:

Recuperar almenys 100 registres.

```python
# seleccionar dataset
d = datasets[6]

# agafar primer resource amb datastore
resource_id = next(r["id"] for r in d["resources"] if r.get("datastore_active"))

# consulta
url = "https://opendata-ajuntament.barcelona.cat/data/api/action/datastore_search"
params = {"resource_id": resource_id, "limit": 100}

data = requests.get(url, params=params).json()

print(len(data["result"]["records"]))
```

### 4. Converteix els resultats a un DataFrame.

```python
records = data["result"]["records"]

df = pd.DataFrame(records)

display(df.head())
```

```python
df.to_csv('./data/emigrants.csv')
```
